```python
import os, shutil
print('=== DISK SPACE ===')
for path in ['/tmp', '/kaggle/working']:
    total, used, free = shutil.disk_usage(path)
    print(f'{path}: free {free/1e9:.1f} GB / total {total/1e9:.1f} GB')
print('\n=== GPU / CUDA ===')
import torch
print('PyTorch:', torch.__version__)
print('CUDA available:', torch.cuda.is_available())
print('GPU count:', torch.cuda.device_count())
for i in range(torch.cuda.device_count()):
    p = torch.cuda.get_device_properties(i)
    print(f'GPU {i}: {p.name} | VRAM {p.total_memory/1e9:.1f} GB')
print('\nSAFE CHECK COMPLETE. No packages installed. No models downloaded.')
```
