## Шаг 1. Безопасная проверка ресурсов
Запусти только следующую ячейку. Она ничего не устанавливает и не скачивает.

```python
import shutil, sys
print('Python:', sys.version.split()[0])
for path in ['/kaggle/working', '/tmp']:
    try:
        total, used, free = shutil.disk_usage(path)
        print(f'{path}: свободно {free/1e9:.1f} GB из {total/1e9:.1f} GB')
    except Exception as e:
        print(path, 'недоступен:', e)
try:
    import torch
    print('PyTorch:', torch.__version__)
    print('CUDA доступна:', torch.cuda.is_available())
    print('Количество GPU:', torch.cuda.device_count())
    for i in range(torch.cuda.device_count()):
        p = torch.cuda.get_device_properties(i)
        print(f'GPU {i}: {p.name}; VRAM {p.total_memory/1e9:.1f} GB')
except Exception as e:
    print('Проверка GPU:', repr(e))
print('Проверка завершена. Ничего не скачивалось и не запускалось в фоне.')
```
