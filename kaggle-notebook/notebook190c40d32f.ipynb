# Wan 2.2 в Kaggle: проверка оборудования
Этот этап проверяет GPU, память, интернет и доступность файлов модели. Большую модель пока не скачиваем.

```python
import os, sys, platform, shutil, subprocess, json, urllib.request
print('Python:', sys.version.split()[0])
print('Платформа:', platform.platform())
print('Свободно на диске, ГБ:', round(shutil.disk_usage('/kaggle/working').free/1024**3, 1))
try:
    import torch
    print('PyTorch:', torch.__version__)
    print('CUDA доступна:', torch.cuda.is_available())
    print('GPU:', torch.cuda.device_count())
    for i in range(torch.cuda.device_count()):
        p = torch.cuda.get_device_properties(i)
        print(f'GPU {i}: {p.name}; память: {p.total_memory/1024**3:.1f} ГБ')
except Exception as e:
    print('Ошибка PyTorch:', repr(e))
```

```python
import urllib.request, json
url = 'https://huggingface.co/api/models/Wan-AI/Wan2.2-TI2V-5B-Diffusers'
try:
    with urllib.request.urlopen(url, timeout=20) as r:
        info = json.loads(r.read())
    print('Hugging Face доступен: ДА')
    print('Модель:', info.get('id'))
    print('Последнее обновление:', info.get('lastModified'))
except Exception as e:
    print('Hugging Face доступен: НЕТ')
    print(type(e).__name__ + ':', str(e)[:400])
```
